```python
def gt(mass_O, dist_to_O):
    mass_nu = len(dist_to_O)
    l = len(np.where(dist_to_O < 2)[0])
    my_lambda = np.minimum(mass_O, l)

    dist_to_O.sort()
    
    gt = np.abs(mass_O - l)
    gt += mass_nu - l
    gt += np.sum(dist_to_O[0:int(np.floor(my_lambda))])
    if len(dist_to_O) > np.floor(my_lambda):
        gt += (my_lambda - np.floor(my_lambda))*dist_to_O[int(np.floor(my_lambda)) + 1]
    gt /= min(mass_O, mass_nu)
    return gt
```
