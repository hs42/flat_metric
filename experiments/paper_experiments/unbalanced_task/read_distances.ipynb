```python
import numpy as np
import pandas as pd
import os
import json
import sys
sys.path.append( '/media/hjs/Data_storage/cluster_27-5-2024/flat_metric/experiments' )
import get_flat_distance

data_p_to_consider = 50
```

```python
directory = '/media/hjs/Data_storage/flat_metric_output/out/unbalanced_task/training'

dirs = os.listdir(directory)


for dir in dirs:
    with open(os.path.join(directory, dir, 'logs{s}config.json'.format(s=os.sep))) as f:
        data = json.load(f)
    type1 = os.path.basename(data['distrib1']['path_train'])
    type2 = os.path.basename(data['distrib2']['path_train'])
    dim = data['distrib2']['dim']
    n = int(data['distrib2']['sample_size'])
    m = int(data['distrib1']['sample_size'])
    distance_train = np.loadtxt(os.path.join(directory, dir, 'logs/train_loss_W.log'), skiprows=1, delimiter=',')
    distance_estimate, distance_std = get_flat_distance.return_distance(distance_train, dim, n/m, data_p_to_consider)
    
    print('Flat distance between', type1, 'and', type2, f'is roughly {distance_estimate:.2f}')
```

```python
directory_Wasserstein = '/media/hjs/Data_storage/flat_metric_output/out/unbalanced_task_W/training'
directory = directory_Wasserstein
dirs = os.listdir(directory)


for dir in dirs:
    with open(os.path.join(directory, dir, 'logs{s}config.json'.format(s=os.sep))) as f:
        data = json.load(f)
    type1 = os.path.basename(data['distrib1']['path_train'])
    type2 = os.path.basename(data['distrib2']['path_train'])
    dim = data['distrib2']['dim']
    n = int(data['distrib2']['sample_size'])
    m = int(data['distrib1']['sample_size'])
    distance_train = np.loadtxt(os.path.join(directory, dir, 'logs/train_loss.log'), skiprows=1, delimiter=',')
    distance_estimate, distance_std = -np.mean(distance_train[-data_p_to_consider:-1, 1]), 0
    
    print('Wasserstein distance between', type1, 'and', type2, f'is roughly {distance_estimate:.2f}')
```
